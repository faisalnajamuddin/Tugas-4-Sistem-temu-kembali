```python
# =============================================
# TUGAS 4 - NO. 1: PREPROCESSING TEKS
# Nama: Abdul Rahman
# NIM : 240210502033
# =============================================

import re
import pandas as pd
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

# Inisialisasi Stemmer & Stopword Sastrawi
stemmer_factory = StemmerFactory()
stemmer = stemmer_factory.create_stemmer()

stopword_factory = StopWordRemoverFactory()
stopword_remover = stopword_factory.create_stop_word_remover()

print("Library berhasil di-import.")
print("Stemmer & Stopword Remover Sastrawi siap digunakan.")
```

```python
# Dataset: 5 dokumen berbahasa Indonesia (teks masih kotor)
dokumen = [
    """Sistem temu kembali informasi adalah sistem yang digunakan untuk menemukan
    dokumen yang relevan berdasarkan query pengguna. Sistem ini banyak digunakan
    di mesin pencari seperti Google dan Bing.""",

    """Jaringan komputer adalah kumpulan komputer yang saling terhubung satu sama
    lain. Jaringan ini memungkinkan komunikasi data antar komputer, seperti
    mengirim email, file, dan video.""",

    """Kecerdasan buatan atau Artificial Intelligence (AI) adalah teknologi yang
    memungkinkan komputer untuk belajar dan mengambil keputusan seperti manusia.
    Contohnya adalah chatbot, mobil otonom, dan rekomendasi produk.""",

    """Pemrograman Python adalah bahasa pemrograman yang populer untuk data science
    dan machine learning. Python memiliki banyak library seperti NumPy, Pandas,
    dan Scikit-learn.""",

    """Basis data adalah kumpulan data yang terorganisir dan disimpan secara
    sistematis. Basis data digunakan untuk menyimpan informasi seperti data
    mahasiswa, nilai, dan jadwal kuliah."""
]

# Tampilkan dokumen asli
for i, doc in enumerate(dokumen, 1):
    print(f"--- Dokumen {i} ---")
    print(doc.strip())
    print()
```

```python
# =============================================
# Fungsi preprocessing_text sesuai instruksi:
# 1. Case folding
# 2. Cleaning (hapus angka, tanda baca, karakter khusus)
# 3. Tokenisasi
# 4. Stopwords removal (Sastrawi)
# 5. Stemming (Sastrawi)
# =============================================

def preprocess_text(text):
    # 1. Case Folding
    text = text.lower()

    # 2. Cleaning: hapus angka, tanda baca, dan karakter khusus
    text = re.sub(r'\d+', '', text)              # hapus angka
    text = re.sub(r'[^\w\s]', ' ', text)         # hapus tanda baca
    text = re.sub(r'\s+', ' ', text).strip()     # rapikan spasi

    # 3. Tokenisasi
    tokens = text.split()

    # 4. Stopwords Removal (Sastrawi)
    text_no_stopword = stopword_remover.remove(text)
    tokens_no_stopword = text_no_stopword.split()

    # 5. Stemming (Sastrawi)
    tokens_stemmed = [stemmer.stem(token) for token in tokens_no_stopword]

    return {
        'case_folding': text,
        'tokens_awal': tokens,
        'tokens_no_stopword': tokens_no_stopword,
        'tokens_stemmed': tokens_stemmed,
        'teks_final': ' '.join(tokens_stemmed)
    }

# Uji fungsi pada dokumen pertama
hasil_uji = preprocess_text(dokumen[0])
print("Uji fungsi preprocess_text() pada Dokumen 1:")
print("Case Folding :", hasil_uji['case_folding'][:80], "...")
print("Token Awal   :", hasil_uji['tokens_awal'][:10], "...")
print("Tanpa Stopword:", hasil_uji['tokens_no_stopword'][:10], "...")
print("Stemmed      :", hasil_uji['tokens_stemmed'][:10], "...")
print("Teks Final   :", hasil_uji['teks_final'][:100], "...")
```

```python
# Terapkan fungsi ke semua dokumen
hasil_preprocessing = [preprocess_text(doc) for doc in dokumen]

# Tampilkan perbandingan sebelum & sesudah (minimal 2 dokumen lengkap)
for i in range(2):  # tampilkan 2 dokumen lengkap
    print(f"==========================================")
    print(f"DOKUMEN {i+1}")
    print(f"==========================================")
    print("SEBELUM PREPROCESSING:")
    print(dokumen[i].strip())
    print("\nSESUDAH PREPROCESSING:")
    print(hasil_preprocessing[i]['teks_final'])
    print()
```

```python
# Hitung jumlah token sebelum & sesudah, serta persentase pengurangan
data_stat = []
for i, (doc, hasil) in enumerate(zip(dokumen, hasil_preprocessing), 1):
    jml_awal = len(hasil['tokens_awal'])
    jml_akhir = len(hasil['tokens_stemmed'])
    pengurangan = ((jml_awal - jml_akhir) / jml_awal) * 100

    data_stat.append({
        'Dokumen': f'Dok {i}',
        'Token Sebelum': jml_awal,
        'Token Sesudah': jml_akhir,
        'Pengurangan (%)': round(pengurangan, 2)
    })

df_stat = pd.DataFrame(data_stat)
print("Tabel Statistik Token:")
print(df_stat.to_string(index=False))
```
