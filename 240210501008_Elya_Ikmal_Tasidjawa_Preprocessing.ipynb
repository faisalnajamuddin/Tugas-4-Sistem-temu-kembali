```python
import re

from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
```

```python
dokumen = [
    "Survei APJII menunjukkan penetrasi internet di Indonesia terus meningkat. Penggunaan internet juga berkaitan dengan pola akses digital dan perkembangan layanan internet di masyarakat.",
    "Perkembangan kecerdasan buatan mendorong organisasi untuk memperkuat sistem keamanan siber. Pembaruan sistem dan pengujian pertahanan diperlukan untuk menghadapi ancaman serangan digital.",
    "Penguatan talenta digital menjadi faktor penting untuk meningkatkan daya saing Indonesia dalam perkembangan teknologi global. Kemampuan sumber daya manusia dibutuhkan untuk menghadapi perubahan industri digital.",
    "Penguasaan keterampilan kecerdasan buatan dapat meningkatkan daya saing tenaga kerja di era digital. Keterampilan tersebut semakin dibutuhkan seiring berkembangnya kebutuhan industri terhadap talenta digital.",
    "Program pengembangan talenta digital membantu mahasiswa meningkatkan kompetensi dan kemampuan teknologi. Talenta digital dari berbagai daerah memiliki peluang untuk berkontribusi dalam perkembangan ekosistem digital Indonesia."
]
```

```python
# Membuat daftar stopwords
stopword_factory = StopWordRemoverFactory()
stopwords = set(stopword_factory.get_stop_words())

# Membuat stemmer
stemmer_factory = StemmerFactory()
stemmer = stemmer_factory.create_stemmer()
```

```python
def preprocess_text(text):
    # 1. Case Folding
    text = text.lower()

    # 2. Cleaning
    text = re.sub(r'[^a-z\s]', ' ', text)

    # 3. Tokenisasi
    tokens = text.split()

    # 4. Stopwords Removal
    tokens = [word for word in tokens if word not in stopwords]

    # 5. Stemming
    tokens = [stemmer.stem(word) for word in tokens]

    return tokens
```

```python
hasil_preprocessing = []

for teks in dokumen:
    hasil = preprocess_text(teks)
    hasil_preprocessing.append(hasil)

for i in range(len(dokumen)):
    print(f"Dokumen {i+1}")
    print("Sebelum :", dokumen[i])
    print("Sesudah :", " ".join(hasil_preprocessing[i]))
    print()
```

```python
statistik = []

for i in range(len(dokumen)):
    # Menghitung token sebelum preprocessing
    token_sebelum = re.findall(r'[a-z]+', dokumen[i].lower())

    # Mengambil token setelah preprocessing
    token_setelah = hasil_preprocessing[i]

    # Menghitung jumlah token
    jumlah_sebelum = len(token_sebelum)
    jumlah_setelah = len(token_setelah)

    # Menghitung persentase pengurangan token
    persentase = ((jumlah_sebelum - jumlah_setelah) / jumlah_sebelum) * 100

    statistik.append([
        f"Dokumen {i+1}",
        jumlah_sebelum,
        jumlah_setelah,
        persentase
    ])

# Menampilkan tabel statistik
print(f"{'Dokumen':<12} {'Token Sebelum':<16} {'Token Setelah':<16} {'Pengurangan Token (%)'}")
print("-" * 70)

for data in statistik:
    print(f"{data[0]:<12} {data[1]:<16} {data[2]:<16} {data[3]:.2f}")
```

Analisis: Berdasarkan hasil preprocessing pada lima dokumen, jumlah token mengalami pengurangan setelah melalui proses case folding, cleaning, tokenisasi, stopwords removal, dan stemming. Secara keseluruhan, jumlah token berkurang dari 118 token menjadi 91 token atau sekitar 22,88%. Pengurangan ini terjadi karena kata-kata umum yang kurang memberikan informasi penting dihapus, sedangkan kata berimbuhan diubah menjadi bentuk dasarnya. Dengan demikian, preprocessing membuat data teks menjadi lebih ringkas dan seragam sehingga dapat membantu sistem IR dalam melakukan pencocokan kata serta pencarian dokumen secara lebih efektif.
