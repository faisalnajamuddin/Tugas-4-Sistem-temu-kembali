```python
import re
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

dokumen = [
    "Sistem Informasi Akademik Fakultas Teknik UNM kini telah terintegrasi dengan basis data terpusat!",
    "Mahasiswa dapat mengakses jaringan komputer kampus melalui wifi yang tersedia di 25 titik area kampus.",
    "Pengumuman: Pelatihan Kecerdasan Buatan (AI) akan diselenggarakan pada tanggal 12-14 Oktober 2026.",
    "Laboratorium komputer Fakultas Teknik menambahkan 30 unit komputer baru untuk mendukung praktikum mahasiswa.",
    "Pendidikan berbasis teknologi digital terus dikembangkan agar mahasiswa lebih siap menghadapi dunia kerja."
]

stemmer = StemmerFactory().create_stemmer()
stopword_remover = StopWordRemoverFactory().create_stop_word_remover()

def preprocess_text(text):
    # case folding -> biar "Sistem" dan "sistem" dihitung sebagai term yang sama
    text = text.lower()

    # cleaning -> buang angka, tanda baca, karakter khusus
    text = re.sub(r"[^a-z\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()

    tokens = text.split()  # tokenisasi

    # stopword_remover butuh input string, jadi token digabung dulu lalu dipecah lagi
    text_no_stop = stopword_remover.remove(" ".join(tokens))
    tokens_no_stop = text_no_stop.split()  # stopwords removal

    # stemming -> kembalikan tiap token ke kata dasar (misal "mengakses" -> "akses")
    return [stemmer.stem(t) for t in tokens_no_stop]

hasil = [preprocess_text(d) for d in dokumen]
jumlah_sebelum = [len(d.split()) for d in dokumen]  # token kasar (belum dibersihkan)
jumlah_sesudah = [len(h) for h in hasil]

for i, (doc, h) in enumerate(zip(dokumen, hasil), start=1):
    print(f"Dokumen {i}")
    print(f"Sebelum : {doc}")
    print(f"Sesudah : {h}\n")

total_sebelum, total_sesudah = sum(jumlah_sebelum), sum(jumlah_sesudah)
persen_reduksi = (1 - total_sesudah / total_sebelum) * 100
print(f"Total token: {total_sebelum} -> {total_sesudah} ({persen_reduksi:.2f}% berkurang)")

#analisis: Hasil preprocessing menunjukkan bahwa teks menjadi lebih bersih dan sederhana setelah melalui case folding, cleaning, 
# tokenisasi, stopword removal, dan stemming. Jumlah token juga berkurang karena kata yang tidak diperlukan, angka, dan tanda baca dihilangkan. 
# Proses ini membuat teks lebih terstruktur dan siap digunakan untuk tahap pembobotan pada Sistem Temu Kembali Informasi.
```
