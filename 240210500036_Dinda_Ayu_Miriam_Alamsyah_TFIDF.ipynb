```python
#Cell 1 Dataset dan preprocessing sederhana

# Mengimpor library yang dibutuhkan
import re
import math
import pandas as pd
from collections import Counter

# Mengimpor Sastrawi untuk stopwords
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

# Membuat daftar stopwords Bahasa Indonesia
stopword_factory = StopWordRemoverFactory()
stopwords = set(stopword_factory.get_stop_words())

# Dataset 4 dokumen bertema komputer dan teknologi
dokumen_tfidf = [
    "Sistem komputer menggunakan perangkat keras dan perangkat lunak.",
    "Jaringan komputer menghubungkan perangkat untuk bertukar data.",
    "Kecerdasan buatan menggunakan algoritma untuk mengolah data.",
    "Sistem temu kembali mencari informasi dari kumpulan dokumen."
]

# Fungsi preprocessing sederhana
def preprocessing_sederhana(text):
    # Case folding
    text = text.lower()

    # Cleaning tanda baca
    text = re.sub(r'[^a-z\s]', ' ', text)

    # Tokenisasi
    tokens = text.split()

    # Stopwords removal
    tokens = [word for word in tokens if word not in stopwords]

    return tokens

# Menerapkan preprocessing ke seluruh dokumen
tokens_dokumen = [
    preprocessing_sederhana(doc)
    for doc in dokumen_tfidf
]

# Menampilkan hasil preprocessing
for i, tokens in enumerate(tokens_dokumen, start=1):
    print(f"Dokumen {i} : {tokens}")
```

```python
#Cell 2 Bag-of-Words dan Raw TF

# Membuat vocabulary dari seluruh dokumen
vocabulary = sorted(set(
    word for tokens in tokens_dokumen for word in tokens
))

# Menghitung jumlah kemunculan setiap term pada setiap dokumen
bow = []

for tokens in tokens_dokumen:
    hitung_kata = Counter(tokens)
    bow.append([hitung_kata.get(term, 0) for term in vocabulary])

# Membuat tabel Bag-of-Words
df_bow = pd.DataFrame(
    bow,
    columns=vocabulary,
    index=["Dokumen 1", "Dokumen 2", "Dokumen 3", "Dokumen 4"]
)

# Raw Count sekaligus menjadi nilai TF
print("Bag-of-Words / Raw TF")
display(df_bow)
```

```python
#Cell 3 DF, IDF, dan TF-IDF manual

# Menghitung Document Frequency (DF)
df_values = {}

for term in vocabulary:
    df_values[term] = sum(
        term in tokens for tokens in tokens_dokumen
    )

# Menghitung IDF menggunakan rumus IDF = log10(N / DF)
N = len(dokumen_tfidf)

idf_values = {
    term: math.log10(N / df)
    for term, df in df_values.items()
}

# Menghitung TF-IDF = TF × IDF
tfidf_manual = df_bow.astype(float).copy()

for term in vocabulary:
    tfidf_manual[term] = df_bow[term] * idf_values[term]

# Menampilkan hasil DF dan IDF
print("Document Frequency (DF)")
display(pd.DataFrame([df_values], index=["DF"]))

print("Inverse Document Frequency (IDF)")
display(pd.DataFrame([idf_values], index=["IDF"]).round(4))

# Menampilkan matriks TF-IDF manual
print("TF-IDF Manual")
display(tfidf_manual.round(4))
```
