```python
import re
import pandas as pd
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
```

```python
stemmer = StemmerFactory().create_stemmer()
stopword = StopWordRemoverFactory().create_stop_word_remover()

dokumen = [
    "PENGUMUMAN! Pendaftaran ujian e-skripsi untuk mahasiswa JTIK Fakultas Teknik UNM angkatan 2026 telah dibuka. Harap segera mendaftar!",
    "Teknologi kecerdasan buatan dan jaringan syaraf tiruan semakin berkembang pesat di tahun 2026 ini, mencapai akurasi 99%.",
    "Mahasiswa diharapkan mengumpulkan tugas Sistem Temu Kembali Informasi dalam format .ipynb melalui platform e-learning paling lambat pukul 23:59.",
    "Laboratorium komputer di kampus sedang dalam tahap perbaikan jaringan lokal (LAN) untuk meningkatkan kecepatan akses internet mahasiswa.",
    "Seminar nasional mengenai Data Mining dan Pengolahan Citra Digital akan diselenggarakan pada hari Senin di aula fakultas teknik."
]
```

```python
def preprocess_text(text):
    text_lower = text.lower()
    
    token_awal = text_lower.split()
    
    text_clean = re.sub(r'[^a-z\s]', ' ', text_lower)
    text_clean = re.sub(r'\s+', ' ', text_clean).strip()
    
    text_stop = stopword.remove(text_clean)
    
    text_stem = stemmer.stem(text_stop)
    
    token_akhir = text_stem.split()
    
    return token_awal, token_akhir, text_stem
```

```python
data_hasil = []

print("=== PERBANDINGAN SEBELUM DAN SESUDAH PREPROCESSING ===\n")

for i, doc in enumerate(dokumen):
    token_awal, token_akhir, teks_bersih = preprocess_text(doc)
    
    jml_awal = len(token_awal)
    jml_akhir = len(token_akhir)
    reduksi = ((jml_awal - jml_akhir) / jml_awal) * 100 if jml_awal > 0 else 0
    
    data_hasil.append({
        'Dokumen': f"Dokumen {i+1}",
        'Jumlah Token Awal': jml_awal,
        'Jumlah Token Akhir': jml_akhir,
        'Pengurangan (%)': f"{reduksi:.2f}%"
    })
    
    if i < 2:
        print(f"--- DOKUMEN {i+1} ---")
        print(f"Sebelum : {doc}")
        print(f"Sesudah : {teks_bersih}\n")

df_stats = pd.DataFrame(data_hasil)
print("=== STATISTIK REDUKSI TOKEN ===")
display(df_stats)
```

## Analisis Preprocessing
Preprocessing teks memberikan dampak yang sangat signifikan terhadap peningkatan kualitas data pada sistem Information Retrieval (IR). Dengan diterapkannya *case folding*, pembersihan karakter non-alfabet, *stopwords removal*, dan *stemming*, noise atau variasi bentuk kata yang tidak informatif berhasil dihilangkan. Hal ini mengakibatkan terjadinya penurunan jumlah token secara signifikan (reduksi data), yang mana secara langsung mengurangi ukuran indeks database, mempercepat proses komputasi, serta meningkatkan akurasi pencocokan (*matching*) antara kata kunci pencarian (*query*) dengan dokumen target.
