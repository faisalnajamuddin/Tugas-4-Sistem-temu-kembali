```python
import re
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

# Inisialisasi Sastrawi (dilakukan di luar fungsi agar performa lebih cepat)
stopword_factory = StopWordRemoverFactory()
stopword_remover = stopword_factory.create_stop_word_remover()

stemmer_factory = StemmerFactory()
stemmer = stemmer_factory.create_stemmer()
```

```python
# 1. Fungsi Preprocessing
def preprocess_text(text):
    # a. Case folding (huruf kecil semua)
    text = text.lower()
    
    # b. Cleaning (menghapus angka, tanda baca, dan karakter khusus)
    text = re.sub(r'[^a-z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    
    # c. Tokenisasi (sementara kita split menjadi list, tapi Sastrawi menerima string)
    # Kita gabungkan kembali untuk diproses Sastrawi, lalu ditokenisasi ulang di akhir
    
    # d. Stopwords removal menggunakan Sastrawi
    text_no_stopwords = stopword_remover.remove(text)
    
    # e. Stemming menggunakan Sastrawi
    text_stemmed = stemmer.stem(text_no_stopwords)
    
    # Tokenisasi akhir (mengembalikan list of tokens)
    tokens = text_stemmed.split()
    return tokens
```

```python
# 2. Dataset 5 Dokumen
dokumen = [
    "Fakultas Teknik UNM meluncurkan sistem pencarian dokumen cerdas berbasis Artificial Intelligence pada tahun 2023!",
    "Mahasiswa diwajibkan untuk mengumpulkan laporan praktikum jaringan komputer ke dalam sistem e-learning paling lambat tanggal 15.",
    "Penelitian ini membahas tentang optimasi algoritma temu kembali informasi pada database berukuran sangat besar.",
    "Sistem pakar dapat digunakan untuk mendiagnosis kerusakan pada perangkat keras komputer dan jaringan.",
    "Pengumuman: Seluruh dosen dan mahasiswa harap menghadiri seminar nasional teknologi informasi di aula rektorat."
]
```

```python
# 3 & 4. Proses dan Tampilkan Hasil serta Perhitungan Token
total_token_sebelum = 0
total_token_sesudah = 0

print("=== HASIL PREPROCESSING ===")
for i, doc in enumerate(dokumen):
    # Hitung token sebelum (tokenisasi sederhana berdasarkan spasi)
    token_sebelum = doc.split()
    jml_sebelum = len(token_sebelum)
    total_token_sebelum += jml_sebelum
    
    # Proses teks
    token_sesudah = preprocess_text(doc)
    jml_sesudah = len(token_sesudah)
    total_token_sesudah += jml_sesudah
    
    # Tampilkan 2 dokumen pertama secara lengkap
    if i < 2:
        print(f"\nDokumen {i+1}:")
        print(f"Sebelum      : {doc}")
        print(f"Sesudah      : {token_sesudah}")
        print(f"Token Sebelum: {jml_sebelum} | Token Sesudah: {jml_sesudah}")

# Hitung Persentase Pengurangan
persentase_pengurangan = ((total_token_sebelum - total_token_sesudah) / total_token_sebelum) * 100

print(f"\n=== STATISTIK TOKEN ===")
print(f"Total Token Sebelum : {total_token_sebelum}")
print(f"Total Token Sesudah : {total_token_sesudah}")
print(f"Persentase Reduksi  : {persentase_pengurangan:.2f}%")
```
