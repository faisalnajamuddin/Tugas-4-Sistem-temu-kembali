```python
#import library re dan sastrawi
import re

from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
```

```python
#menyiapkan sastrawi
stopword_factory = StopWordRemoverFactory()
stopword_remover = stopword_factory.create_stop_word_remover()

stemmer_factory = StemmerFactory()
stemmer = stemmer_factory.create_stemmer()
```

```python
#membuat 5 dokumen
dokumen = [
    "Mahasiswa menggunakan komputer untuk mengerjakan tugas kuliah.",
    "Pendidikan yang merata dan terjangkau merupakan pilar utama dalam membangun sumber daya manusia yang berkualitas",
    "Artificial Inteligence merupakan kumpulan teknologi yang memiliki sistem adaptif untuk belajar sendiri dari data.",
    "Teknik Komputer adalah bidang ilmu yang menggabungkan hardware dan software untuk merancang sistem komputer dan jaringan.",
    "Global Digital Compact adalah kesepakatan dunia untuk memperkuat kerja sama dalam mengelola teknologi digital dan AI."
]

for i, teks in enumerate(dokumen, 1):
    print("Dokumen",  i)
    print(teks)
    print
```

```python
#fungsi preprocessing
def preprocess_text(text):
    #case folding
    text = text.lower()

    #cleaning
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'[^a-z\s]', '', text)

    #tokenisasi
    tokens = text.split()

    #stopwords removal
    text = ' '.join(tokens)
    text = stopword_remover.remove(text)

    tokens = text.split()

    #stemming
    tokens = [stemmer.stem(token) for token in tokens]

    return tokens
```

```python
contoh = dokumen[0][1]

print("Sebelum:")
print(contoh)

print("\nSesudah:")
print(preprocess_text(contoh))
```

```python
#preprocessing
hasil_preprocessing = []

for teks in dokumen:
    hasil = preprocess_text(teks)
    hasil_preprocessing.append(hasil)

for i, hasil in enumerate(hasil_preprocessing, 1):
    print("Dokumen", i)
    print(hasil)
    print()
```

```python
#tampilkan perbandingan
for i in range(len(dokumen)):
    print("=" * 50)
    print("Dokumen", i + 1)

    print("\nSebelum preprocessing:")
    print(dokumen[i])

    print("\nSesudah preprocessing:")
    print(hasil_preprocessing[i])
```

```python
#menghitung token sebelum dan sesudah proprocessing

print("Sebelum preprocessing :")
jumlah_token_sebelum = []

for teks in dokumen:
    teks_bersih = re.sub(r'[^a-zA-A\s]', '', teks)
    tokens = teks_bersih.split()

    jumlah_token_sebelum.append(len(tokens))

for i, jumlah in enumerate(jumlah_token_sebelum, 1):
    print("Dokumen", i, ":", jumlah, "token")

print()
print("Setelah preprocessing :")
jumlah_token_sesudah = []

for hasil in hasil_preprocessing:
    jumlah_token_sesudah.append(len(hasil))

for i, jumlah in enumerate(jumlah_token_sesudah, 1):
    print("Dokumen", i, ":", jumlah, "token")
```

```python
#persentase pengurangan token
persentase_pengurangan = []

for sebelum, sesudah in zip(jumlah_token_sebelum, jumlah_token_sesudah):
    hasil = ((sebelum - sesudah) / sebelum)*100
    persentase_pengurangan.append(hasil)

for i, hasil in enumerate(persentase_pengurangan, 1):
    print("Dokumen", i, ":", round(hasil, 2), "%")
```

```python
#tabel hasil

import pandas as pd

hasil_tabel = pd.DataFrame({
    "Dokumen": ["Dokumen" + str(i) for i in range(1, 6)],
    "Token Sebelum": jumlah_token_sebelum,
    "Token Sesudah": jumlah_token_sesudah,
    "Pengurangan (%)": [round(x, 2) for x in persentase_pengurangan]
})

hasil_tabel
```

Analisis Singkat:
Dampak preprocessing terhadap kualitas data untuk membuat data teks menjadi lebih bersih dan terstruktur sehingga lebih mudah digunakan dalam sistem temu kembali informasi (IR). Case folding membuat penulisan huruf menjadi seragam, cleaning menghilangkan angka dan tanda baca yang tidak diperlukan, stopwords removal mengurangi kata-kata yang kurang penting, sedangkan stemming mengubah kata berimbuhan menjadi bentuk dasar. Proses tersebut dapat mengurangi jumlah token dan membuat data menjadi lebih sederhana, sehingga sistem dapat lebih fokus pada kata-kata yang dianggap penting dalam pencarian dokumen.
