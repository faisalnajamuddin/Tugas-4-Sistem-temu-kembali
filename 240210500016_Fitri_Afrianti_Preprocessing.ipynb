```python
# ---------- SEL 1: Import Library ----------
!pip install PySastrawi

import re                                                    # regular expression untuk cleaning
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory   # stemmer Bahasa Indonesia
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory  # daftar stopwords

print("Library berhasil di-import.")
```

```python
# ---------- SEL 2: Dataset (5 dokumen, masih kotor) ----------
dokumen = [
    # Dok 1 - pengumuman kampus
    "PENGUMUMAN: Fakultas Teknik UNM akan mengadakan Seminar Nasional Teknologi Informasi pada tanggal "
    "15 Agustus 2025 di Gedung Auditorium! Seluruh mahasiswa diwajibkan mengikuti kegiatan tersebut, "
    "dan pendaftaran dapat dilakukan melalui website resmi fakultas (www.ft.unm.ac.id).",

    # Dok 2 - berita teknologi
    "Perkembangan kecerdasan buatan semakin pesat di tahun 2025. Banyak perusahaan menggunakan "
    "teknologi tersebut untuk mengolah data, mempercepat pekerjaan, dan meningkatkan layanan kepada "
    "pelanggan!!! Namun, para ahli mengingatkan pentingnya etika dalam penggunaan AI.",

    # Dok 3 - komputer
    "Sistem operasi adalah perangkat lunak yang mengatur seluruh sumber daya komputer, seperti "
    "prosesor, memori, dan penyimpanan. Windows, Linux, dan macOS merupakan contoh sistem operasi "
    "yang banyak digunakan oleh 90% pengguna komputer di dunia.",

    # Dok 4 - pendidikan
    "Mahasiswa Teknik Komputer sedang mempelajari pemrograman dan jaringan komputer pada semester 5. "
    "Dosen memberikan tugas praktikum yang harus dikumpulkan paling lambat tanggal 20 September 2025 "
    "melalui e-learning kampus #UNM @teknik.",

    # Dok 5 - sistem temu kembali
    "Sistem temu kembali informasi membantu pengguna menemukan dokumen yang relevan dengan kebutuhan "
    "mereka. Proses pencarian dilakukan dengan mencocokkan kata kunci pada kumpulan dokumen, kemudian "
    "hasilnya diurutkan berdasarkan tingkat kemiripan!",
]

print(f"Jumlah dokumen: {len(dokumen)}")
```

```python
# ---------- SEL 3: Fungsi preprocess_text ----------
# Urutan: case folding -> cleaning -> tokenisasi -> stopwords removal -> stemming

# Inisialisasi Sastrawi (dibuat sekali di luar fungsi agar efisien)
stemmer = StemmerFactory().create_stemmer()
STOPWORDS = set(StopWordRemoverFactory().get_stop_words())   # daftar stopwords bawaan Sastrawi

def preprocess_text(text, return_steps=False):
    """
    Pipeline preprocessing teks Bahasa Indonesia.
    Urutan: case folding -> cleaning -> tokenisasi -> stopwords removal -> stemming.

    Parameter
    ---------
    text         : str  - teks mentah
    return_steps : bool - jika True, kembalikan juga hasil tiap tahap (untuk demonstrasi)

    Return
    ------
    list[str] berisi token hasil akhir (atau dict berisi semua tahap jika return_steps=True)
    """
    # 1. Case folding: semua huruf -> huruf kecil
    folded = text.lower()

    # 2. Cleaning: hapus angka, tanda baca, dan karakter khusus (sisakan huruf a-z dan spasi)
    cleaned = re.sub(r"[^a-z\s]", " ", folded)
    cleaned = re.sub(r"\s+", " ", cleaned).strip()          # rapikan spasi berlebih

    # 3. Tokenisasi: pecah berdasarkan spasi
    tokens = cleaned.split()

    # 4. Stopwords removal: buang token yang ada di daftar stopwords Sastrawi
    no_stop = [t for t in tokens if t not in STOPWORDS]

    # 5. Stemming: ubah tiap token menjadi kata dasar
    stemmed = [stemmer.stem(t) for t in no_stop]

    if return_steps:
        return {"case_folding": folded, "cleaning": cleaned, "tokenisasi": tokens,
                "stopword_removal": no_stop, "stemming": stemmed}
    return stemmed

# Uji singkat pada satu kalimat
contoh = "Mahasiswa-mahasiswa sedang MENGERJAKAN tugas di 2025!"
print("Teks   :", contoh)
print("Hasil  :", preprocess_text(contoh))
```

```python
# ---------- SEL 4: Terapkan pada kelima dokumen ----------
hasil = [preprocess_text(d) for d in dokumen]

for i, tokens in enumerate(hasil, start=1):
    print(f"Dok {i}: {tokens}\n")
```

```python
# ---------- SEL 5: Perbandingan lengkap per tahap (Dokumen 1 dan 2) ----------
def tampilkan_lengkap(nomor):
    """Menampilkan hasil setiap tahap preprocessing untuk satu dokumen."""
    teks = dokumen[nomor - 1]
    s = preprocess_text(teks, return_steps=True)
    garis = "=" * 90
    print(garis)
    print(f"DOKUMEN {nomor}")
    print(garis)
    print("[Teks Asli]\n", teks, "\n")
    print("[1. Case Folding]\n", s["case_folding"], "\n")
    print("[2. Cleaning]\n", s["cleaning"], "\n")
    print(f"[3. Tokenisasi]  ({len(s['tokenisasi'])} token)\n", s["tokenisasi"], "\n")
    print(f"[4. Stopwords Removal]  ({len(s['stopword_removal'])} token)\n", s["stopword_removal"], "\n")
    print(f"[5. Stemming / Hasil Akhir]  ({len(s['stemming'])} token)\n", s["stemming"], "\n")

tampilkan_lengkap(1)
tampilkan_lengkap(2)
```

```python
# ---------- SEL 6: Ringkasan sebelum vs sesudah (semua dokumen) ----------
for i, teks in enumerate(dokumen, start=1):
    print(f"--- Dokumen {i} ---")
    print("SEBELUM :", teks)
    print("SESUDAH :", " ".join(hasil[i-1]))
    print()
```

```python
# ---------- SEL 7: Statistik jumlah token ----------
# Sebelum  = jumlah kata pada teks asli (dipisah spasi)
# Sesudah  = jumlah token hasil akhir preprocessing
# Persen   = (sebelum - sesudah) / sebelum * 100

print(f"{'Dokumen':<10}{'Sebelum':>10}{'Sesudah':>10}{'Berkurang':>12}{'Pengurangan (%)':>18}")
print("-" * 60)

total_sebelum, total_sesudah = 0, 0
for i, teks in enumerate(dokumen, start=1):
    sebelum = len(teks.split())          # token mentah (split berdasarkan spasi)
    sesudah = len(hasil[i-1])            # token setelah preprocessing
    persen  = (sebelum - sesudah) / sebelum * 100
    total_sebelum += sebelum
    total_sesudah += sesudah
    print(f"Dok {i:<6}{sebelum:>10}{sesudah:>10}{sebelum - sesudah:>12}{persen:>17.2f}%")

print("-" * 60)
persen_total = (total_sebelum - total_sesudah) / total_sebelum * 100
print(f"{'TOTAL':<10}{total_sebelum:>10}{total_sesudah:>10}{total_sebelum - total_sesudah:>12}{persen_total:>17.2f}%")
```

Preprocessing meningkatkan kualitas data untuk sistem IR karena mengubah teks mentah yang "kotor" menjadi representasi yang seragam dan lebih informatif. Pada percobaan ini jumlah token turun dari 157 menjadi 103 (berkurang 34,39%), dengan pengurangan per dokumen antara 21,21% sampai 43,75%; artinya sepertiga lebih token adalah stopwords atau noise (mis. yang, dan, di, pada) yang tidak membantu membedakan dokumen. Case folding dan cleaning membuat kata seperti "Teknik" dan "teknik" dihitung sebagai satu term, sedangkan stemming menyatukan bentuk berimbuhan (mis. mengolah, pengolahan → olah), sehingga vocabulary lebih kecil, indeks lebih hemat, dan query bisa cocok dengan dokumen walaupun bentuk katanya berbeda (recall naik). Selain itu, term yang tersisa (mis. komputer, teknologi, informasi, sistem) lebih mewakili topik dokumen sehingga pembobotan seperti TF-IDF menjadi lebih akurat. Namun ada keterbatasan: stemmer Sastrawi kadang melakukan over-stemming (mis. pengumuman → umum, perkembangan → kembang) dan daftar stopwords bawaannya menghapus kata kembali sehingga frasa "temu kembali" menjadi "temu"; selain itu alamat web (www.ft.unm.ac.id) terpecah menjadi token tak bermakna. Jadi preprocessing perlu disesuaikan dengan domain agar tidak menghilangkan informasi penting.
