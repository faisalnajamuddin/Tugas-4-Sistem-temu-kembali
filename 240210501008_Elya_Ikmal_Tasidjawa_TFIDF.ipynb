```python
import re
import math
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
dokumen = [
    "Kecerdasan buatan membantu perkembangan teknologi komputer.",
    "Jaringan komputer menghubungkan perangkat untuk berbagi data.",
    "Teknologi jaringan membantu pertukaran data antar komputer.",
    "Sistem temu kembali membantu pencarian informasi digital."
]

for i, teks in enumerate(dokumen, start=1):
    print(f"Dokumen {i}: {teks}")
```

```python
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

stopword_factory = StopWordRemoverFactory()
stopwords = set(stopword_factory.get_stop_words())
```

```python
def preprocess_text(text):
    # Case Folding
    text = text.lower()

    # Cleaning
    text = re.sub(r'[^a-z\s]', ' ', text)

    # Tokenisasi
    tokens = text.split()

    # Stopwords Removal
    tokens = [word for word in tokens if word not in stopwords]

    return tokens
```

```python
hasil_preprocessing = []

for teks in dokumen:
    hasil = preprocess_text(teks)
    hasil_preprocessing.append(hasil)

for i in range(len(dokumen)):
    print(f"Dokumen {i+1}")
    print("Sebelum :", dokumen[i])
    print("Sesudah :", hasil_preprocessing[i])
    print()
```

```python
# Membuat vocabulary dari seluruh token unik
vocabulary = sorted(set(
    word
    for tokens in hasil_preprocessing
    for word in tokens
))

print("Vocabulary:")
print(vocabulary)
print("\nJumlah term unik:", len(vocabulary))

# Membuat Bag-of-Words (raw count)
bow = []

for tokens in hasil_preprocessing:
    baris = []

    for term in vocabulary:
        jumlah = tokens.count(term)
        baris.append(jumlah)

    bow.append(baris)

# Menampilkan dalam bentuk tabel
df_bow = pd.DataFrame(
    bow,
    index=[f"Dokumen {i+1}" for i in range(len(dokumen))],
    columns=vocabulary
)

df_bow
```

```python
# Menghitung Term Frequency (TF)
tf = []

for tokens in hasil_preprocessing:
    baris_tf = []

    for term in vocabulary:
        frekuensi = tokens.count(term)
        baris_tf.append(frekuensi)

    tf.append(baris_tf)

# Menampilkan TF dalam bentuk tabel
df_tf = pd.DataFrame(
    tf,
    index=[f"Dokumen {i+1}" for i in range(len(dokumen))],
    columns=vocabulary
)

print("Term Frequency (TF):")
df_tf
```

```python
# Menghitung Document Frequency (DF)
df = []

for term in vocabulary:
    jumlah_dokumen = 0

    for tokens in hasil_preprocessing:
        if term in tokens:
            jumlah_dokumen += 1

    df.append(jumlah_dokumen)

# Menampilkan DF dalam bentuk tabel
df_document = pd.DataFrame({
    "Term": vocabulary,
    "DF": df
})

print("Document Frequency (DF):")
df_document
```

```python
# Jumlah seluruh dokumen
N = len(dokumen)

# Menghitung Inverse Document Frequency (IDF)
idf = []

for nilai_df in df:
    nilai_idf = math.log10(N / nilai_df)
    idf.append(nilai_idf)

# Menampilkan DF dan IDF
df_idf = pd.DataFrame({
    "Term": vocabulary,
    "DF": df,
    "IDF": idf
})

print("Document Frequency (DF) dan Inverse Document Frequency (IDF):")
df_idf
```

```python
# Menghitung TF-IDF secara manual
tfidf_manual = []

for baris_tf in tf:
    baris_tfidf = []

    for i in range(len(vocabulary)):
        nilai_tfidf = baris_tf[i] * idf[i]
        baris_tfidf.append(nilai_tfidf)

    tfidf_manual.append(baris_tfidf)

# Menampilkan matriks TF-IDF manual
df_tfidf_manual = pd.DataFrame(
    tfidf_manual,
    index=[f"Dokumen {i+1}" for i in range(len(dokumen))],
    columns=vocabulary
)

print("Matriks TF-IDF Manual:")
df_tfidf_manual
```

```python
dokumen_bersih = [
    " ".join(tokens) for tokens in hasil_preprocessing
]

for i, teks in enumerate(dokumen_bersih, start=1):
    print(f"Dokumen {i}: {teks}")
```

```python
# Inisialisasi vectorizer dengan parameter default
vectorizer = TfidfVectorizer()

# Fit dan transform dokumen
tfidf_matrix = vectorizer.fit_transform(dokumen_bersih)

# Menampilkan vocabulary
print("Vocabulary:")
print(vectorizer.get_feature_names_out())

# Menampilkan matriks TF-IDF sebagai array
print("\nMatriks TF-IDF:")
print(tfidf_matrix.toarray())

# Menampilkan matriks TF-IDF dalam bentuk DataFrame
df_tfidf = pd.DataFrame(
    tfidf_matrix.toarray(),
    index=[f"Dokumen {i+1}" for i in range(len(dokumen))],
    columns=vectorizer.get_feature_names_out()
)

print("\nDataFrame TF-IDF:")
df_tfidf
```

```python
print("TF-IDF Manual:")
display(df_tfidf_manual)

print("\nTF-IDF Scikit-learn:")
display(df_tfidf)
```

Perbandingan: Nilai TF-IDF manual dan Scikit-learn memiliki nilai numerik yang berbeda. Perhitungan manual menggunakan IDF klasik log10(N/df) tanpa normalisasi, sedangkan TfidfVectorizer menggunakan perhitungan default yang menerapkan smoothing dan normalisasi. Meskipun nilai numeriknya berbeda, keduanya tetap memberikan bobot lebih tinggi pada term yang lebih jarang muncul dalam koleksi dokumen.

```python
# Mencari term dengan bobot TF-IDF tertinggi pada setiap dokumen
for i in range(len(df_tfidf)):
    nilai_tertinggi = df_tfidf.iloc[i].max()

    term_tertinggi = df_tfidf.columns[
        df_tfidf.iloc[i] == nilai_tertinggi
    ].tolist()

    print(f"Dokumen {i+1}")
    print("Term tertinggi :", ", ".join(term_tertinggi))
    print(f"Bobot TF-IDF   : {nilai_tertinggi:.6f}")
    print()
```

Analisis: Pada Dokumen 1, term buatan, kecerdasan, dan perkembangan memiliki bobot TF-IDF tertinggi sebesar 0,474771. Pada Dokumen 2, term berbagi, menghubungkan, dan perangkat memiliki bobot tertinggi sebesar 0,463709. Pada Dokumen 3, term pertukaran memiliki bobot tertinggi sebesar 0,521314, sedangkan pada Dokumen 4 term digital, informasi, pencarian, sistem, dan temu memiliki bobot tertinggi sebesar 0,430037. Term-term tersebut memperoleh bobot tinggi karena lebih jarang muncul pada koleksi dokumen, sehingga lebih mencerminkan ciri atau topik khusus dari masing-masing dokumen.
