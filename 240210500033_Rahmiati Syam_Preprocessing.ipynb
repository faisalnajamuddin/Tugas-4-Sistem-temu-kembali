```python
import re
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
```

Dataset yang digunakan disini ada 5 dokumen

```python
documents = {
    "Dokumen 1": "Teknologi komputer membantu mahasiswa menyelesaikan tugas kuliah dengan lebih cepat dan teratur.",
    "Dokumen 2": "Jaringan komputer menghubungkan perangkat agar dapat bertukar data dan informasi secara cepat.",
    "Dokumen 3": "Kecerdasan buatan digunakan untuk mengolah data, mengenali pola, dan membantu pengambilan keputusan.",
    "Dokumen 4": "Sistem temu kembali informasi membantu pengguna menemukan dokumen yang sesuai dengan kebutuhan pencarian.",
    "Dokumen 5": "Keamanan informasi penting untuk melindungi data pengguna dari serangan dan penyalahgunaan teknologi."
}

for nama, teks in documents.items():
    print(f"{nama}: {teks}")
```

membuat fungsi Python bernama preprocessing_text yang melakukan urutan case folding, cleaning, tokenisasi, stopwords removal menggunakan sastrawi, dan stemming menggunakan sastrawi

```python
stopword_factory = StopWordRemoverFactory()
stopword_remover = stopword_factory.create_stop_word_remover()

stemmer_factory = StemmerFactory()
stemmer = stemmer_factory.create_stemmer()


def preprocess_text(text):
    text = text.lower()

    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()

    tokens = text.split()

    text_without_stopwords = stopword_remover.remove(" ".join(tokens))
    tokens_without_stopwords = text_without_stopwords.split()

    stemmed_text = stemmer.stem(" ".join(tokens_without_stopwords))
    final_tokens = stemmed_text.split()

    return final_tokens
```

menerapkan fungsi preprocessing_text pada 5 dokumen

```python
preprocessed_documents = {}

for nama, teks in documents.items():
    hasil = preprocess_text(teks)
    preprocessed_documents[nama] = hasil
    print(f"{nama}: {' '.join(hasil)}")
```

perbandingan sebelum dan sesudah preprocessing dokumen

```python
for nama in ["Dokumen 1", "Dokumen 2"]:
    print("=" * 70)
    print(nama)
    print("Sebelum :", documents[nama])
    print("Sesudah :", " ".join(preprocessed_documents[nama]))
```

menghitung dan menampilkan jumlah token sebelum dan sesudah preprocessing, serta menghitung persentase pengurangan token

```python
results = []

for nama, teks in documents.items():
    # Token sebelum preprocessing: cleaning dasar + tokenisasi
    cleaned = re.sub(r'[^a-zA-Z\s]', ' ', teks.lower())
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    tokens_before = cleaned.split()

    tokens_after = preprocessed_documents[nama]

    jumlah_before = len(tokens_before)
    jumlah_after = len(tokens_after)

    if jumlah_before > 0:
        pengurangan = ((jumlah_before - jumlah_after) / jumlah_before) * 100
    else:
        pengurangan = 0

    results.append(
        (nama, jumlah_before, jumlah_after, round(pengurangan, 2))
    )

print(f"{'Dokumen':<15}{'Token Sebelum':<18}{'Token Sesudah':<18}{'Pengurangan (%)'}")
print("-" * 70)

for row in results:
    print(f"{row[0]:<15}{row[1]:<18}{row[2]:<18}{row[3]}")
```

Analisis
Berdasarkan hasil preprocessing yang sudah dilakukan dapat disimpulkan bahwa proses preproessing cukup signifikan terhadap bentuk dan jumlah teks dari dokumen. pada kondisi awal dokumen masih mengandung kata umum, tanda baca, karakter khusus, serta kata-kata imbuhan. preprocessing membuat data teks menjadi lebih seragam dan relevan untuk sistem temu kembali. case holding mengurangi perbedaan penulisan huruf kapital, cleaning untuk menghilangkan karakter yang tidak diperlukan, tokenisasi memecahkan teks menjadi unit kata, stopwords removal untuk mengurangi kata umum yang kurang informatif, sedangkan stemming dapat mengubah kata berimbuhan ke bentuk dasarnya.
