1. Import library dan dokumen

```python
import re
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

dokumen = [
    "Fakultas Teknik Universitas Negeri Makassar meluncurkan laboratorium kecerdasan buatan baru pada tahun 2026. Laboratorium ini bertujuan untuk meningkatkan kompetensi mahasiswa dalam riset pembelajaran mesin.",
    "Sistem temu kembali informasi sangat krusial dalam pengelolaan arsip digital kampus. Algoritma pencarian yang efisien membantu sivitas akademika menemukan dokumen riset dengan cepat dan akurat.",
    "Mahasiswa Teknik Komputer sedang mengembangkan aplikasi berbasis Internet of Things. Perangkat sensor cerdas tersebut dipasang untuk memantau kualitas udara serta suhu ruangan secara real-time.",
    "Penerapan komputasi awan pada server universitas berhasil mengoptimalkan kinerja sistem informasi akademik. Skalabilitas server meningkat drastis saat melayani ribuan mahasiswa yang mengakses portal KRS.",
    "Keamanan siber menjadi prioritas penting dalam melindungi integritas basis data universitas dari serangan malware. Pihak fakultas mewajibkan penerapan autentikasi ganda bagi seluruh pengguna jaringan internal."
]
```

2. Inisialisasi Sastrawi

```python
stop_factory = StopWordRemoverFactory()
stopwords = stop_factory.get_stop_words()

stemmer_factory = StemmerFactory()
stemmer = stemmer_factory.create_stemmer()
```

3. Definisi fungsi preprocessing

```python
def preprocess_text(text):
    text = text.lower()
    
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
  
    tokens = text.split()
 
    tokens = [token for token in tokens if token not in stopwords]
    
    tokens = [stemmer.stem(token) for token in tokens]
    
    return tokens
```

4. Eksekusi dokumen & evaluasi jumlah token

```python
hasil_preprocessing = []

for i, doc in enumerate(dokumen, 1):
    tokens_sebelum = re.findall(r'\b[a-zA-Z]+\b', doc.lower())
    
    tokens_sesudah = preprocess_text(doc)
    
    hasil_preprocessing.append({
        "Dokumen": i,
        "Sebelum": doc,
        "Sesudah": " ".join(tokens_sesudah),
        "Token Sebelum": len(tokens_sebelum),
        "Token Sesudah": len(tokens_sesudah)
    })
```

5. Tampilkan perbandingan detail

```python
for hasil in hasil_preprocessing[:5]:
    print("=" * 70)
    print(f"DOKUMEN {hasil['Dokumen']}")
    print("Sebelum preprocessing:")
    print(hasil["Sebelum"])
    print("\nSesudah preprocessing:")
    print(hasil["Sesudah"])
    print("=" * 70)
```

6. Tampilkan tabel ringkasan reduksi token untuk seluruh dokumen

```python
print(f"{'Dokumen':<10}{'Token Sebelum':<18}{'Token Sesudah':<18}{'Pengurangan':<15}")
print("-" * 61)

for hasil in hasil_preprocessing:
    sebelum = hasil["Token Sebelum"]
    sesudah = hasil["Token Sesudah"]
    
    pengurangan = ((sebelum - sesudah) / sebelum) * 100 if sebelum > 0 else 0
    
    print(
        f"{hasil['Dokumen']:<10}"
        f"{sebelum:<18}"
        f"{sesudah:<18}"
        f"{pengurangan:.2f}%"
    )
```

penerapan tahapan preprocessing teks bahasa Indonesia terbukti sangat efektif dalam meringkas dan menormalisasi data. Melalui proses case folding, pembersihan karakter, serta penghapusan stopwords dan stemming menggunakan pustaka Sastrawi, sistem berhasil menyaring kata-kata umum pada kelima dokumen yang diuji. Proses ini berdampak langsung pada penurunan jumlah token teks secara persentase, sehingga menghasilkan representasi data yang jauh lebih bersih, padat, dan efisien untuk diproses oleh sistem temu kembali informasi.
