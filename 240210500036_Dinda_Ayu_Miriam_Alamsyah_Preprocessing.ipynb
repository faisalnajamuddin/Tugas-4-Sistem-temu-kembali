```python
#Cell 1 Import library dan siapkan preprocessing

# Mengimpor library yang dibutuhkan
import re
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

# Membuat daftar stopwords Bahasa Indonesia
stopword_factory = StopWordRemoverFactory()
stopwords = set(stopword_factory.get_stop_words())

# Membuat stemmer Bahasa Indonesia
stemmer_factory = StemmerFactory()
stemmer = stemmer_factory.create_stemmer()
```

```python
#Cell 2 Dataset dan fungsi preprocessing

# Membuat 5 dokumen berbahasa Indonesia dengan topik teknologi
dokumen = [
    "Teknologi komputer berkembang sangat pesat. Komputer membantu manusia menyelesaikan pekerjaan dengan cepat.",

    "Jaringan komputer menghubungkan berbagai perangkat untuk bertukar data dan informasi melalui jaringan internet.",

    "Kecerdasan buatan membantu komputer melakukan tugas yang sebelumnya membutuhkan kemampuan manusia.",

    "Sistem informasi membantu organisasi mengelola data menjadi informasi yang berguna untuk pengambilan keputusan.",

    "Pendidikan berbasis teknologi menggunakan komputer dan internet untuk membantu proses pembelajaran mahasiswa dan siswa."
]

# Fungsi untuk melakukan preprocessing teks
def preprocess_text(text):
    # 1. Case folding: mengubah semua huruf menjadi huruf kecil
    text = text.lower()

    # 2. Cleaning: menghapus angka, tanda baca, dan karakter khusus
    text = re.sub(r'[^a-z\s]', ' ', text)

    # 3. Tokenisasi: memisahkan teks menjadi kata
    tokens = text.split()

    # 4. Stopwords removal: menghapus kata umum
    tokens = [word for word in tokens if word not in stopwords]

    # 5. Stemming: mengubah kata ke bentuk dasarnya
    tokens = [stemmer.stem(word) for word in tokens]

    return tokens
```

```python
#Cell 3 Tampilkan sebelum dan sesudah preprocessing

# Menampilkan hasil preprocessing untuk seluruh dokumen
for i, teks in enumerate(dokumen, start=1):
    hasil = preprocess_text(teks)

    print(f"Dokumen {i}")
    print("Sebelum :", teks)
    print("Sesudah :", " ".join(hasil))
    print("-" * 70)
```

```python
#Cell 4 Menghitung pengurangan token

# Membandingkan jumlah token sebelum dan sesudah preprocessing
print("HASIL PERBANDINGAN TOKEN")
print("=" * 70)

for i, teks in enumerate(dokumen, start=1):
    # Token sebelum preprocessing
    token_sebelum = re.findall(r'\b[a-zA-Z]+\b', teks.lower())

    # Token setelah preprocessing
    token_sesudah = preprocess_text(teks)

    jumlah_sebelum = len(token_sebelum)
    jumlah_sesudah = len(token_sesudah)

    # Menghitung persentase pengurangan token
    pengurangan = ((jumlah_sebelum - jumlah_sesudah) / jumlah_sebelum) * 100

    print(f"Dokumen {i}")
    print(f"Token sebelum : {jumlah_sebelum}")
    print(f"Token sesudah : {jumlah_sesudah}")
    print(f"Pengurangan   : {pengurangan:.2f}%")
    print("-" * 50)
```

```python
### Analisis Cell 1
# Pada cell ini dilakukan import library yang diperlukan untuk proses preprocessing teks. Library `re` digunakan untuk membersihkan teks dari karakter yang tidak diperlukan, sedangkan Sastrawi digunakan untuk menghapus stopword dan melakukan stemming pada Bahasa Indonesia. Stopword remover digunakan untuk menghilangkan kata-kata umum yang kurang memberikan informasi, sedangkan stemmer digunakan untuk mengubah kata berimbuhan menjadi bentuk dasarnya.

### Analisis Cell 2
# Pada cell ini dibuat lima dokumen yang digunakan sebagai data uji dengan topik teknologi. Selanjutnya dibuat fungsi `preprocess_text()` yang menjalankan beberapa tahapan preprocessing, yaitu case folding untuk mengubah huruf menjadi huruf kecil, cleaning untuk menghapus tanda baca dan karakter selain huruf, tokenisasi untuk memisahkan teks menjadi kata, stopwords removal untuk menghilangkan kata yang kurang penting, dan stemming untuk mengubah kata ke bentuk dasarnya. Hasil dari fungsi ini berupa kumpulan token yang sudah melalui seluruh tahapan preprocessing.

### Analisis Cell 3
# Pada cell ini fungsi preprocessing diterapkan pada seluruh dokumen yang telah dibuat. Program menampilkan kondisi teks sebelum dan sesudah preprocessing sehingga dapat dilihat perubahan yang terjadi. Setelah preprocessing, teks menjadi lebih sederhana karena huruf sudah diseragamkan, tanda baca dihilangkan, stopword dikurangi, dan kata berimbuhan diubah ke bentuk dasarnya.

### Analisis Cell 4
# Pada cell ini dilakukan perbandingan jumlah token sebelum dan sesudah preprocessing. Jumlah token sebelum preprocessing diperoleh dari teks asli, sedangkan jumlah token sesudah preprocessing diperoleh dari hasil fungsi `preprocess_text()`. Selanjutnya dihitung persentase pengurangan token untuk mengetahui seberapa banyak data teks yang berkurang setelah proses preprocessing. Semakin banyak token yang dihilangkan, menunjukkan bahwa proses preprocessing berhasil mengurangi kata atau karakter yang dianggap kurang diperlukan dalam pengolahan teks.
```
